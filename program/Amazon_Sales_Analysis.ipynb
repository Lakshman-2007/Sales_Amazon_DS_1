```python
daily_sales=df.groupby("Order Date")["Sales"].sum().sort_index()
```

```python
print(daily_sales.head())
```

```python
print(daily_sales.tail())
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(15,9))
plt.plot(daily_sales.index,daily_sales.values)
plt.title("Daily Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid()
plt.plot()
```

```python
weekly_sales = daily_sales.resample("W").sum()

weekly_sales.plot(figsize=(15, 5), title="Weekly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Weekly Sales")
plt.grid()
plt.show()
```

```python
rolling_sales = daily_sales.rolling(window=30).mean()

plt.figure(figsize=(15, 5))
plt.plot(daily_sales, label="Daily Sales")
plt.plot(rolling_sales, label="30-Day Rolling Average")
plt.title("Daily Sales vs 30-Day Rolling Average")
plt.legend()
plt.grid()
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

result = adfuller(daily_sales)

print("ADF Statistic:", result[0])
print("p-value:", result[1])
```

```python
daily_sales = daily_sales.asfreq("D", fill_value=0)

print(daily_sales.index.freq)
print(daily_sales.isnull().sum())
```

```python
sales_diff = daily_sales.diff().dropna()
```

```python
plt.figure(figsize=(15, 5))
plt.plot(sales_diff)
plt.title("First Order Differencing")
plt.xlabel("Date")
plt.ylabel("Difference in Sales")
plt.grid()
plt.show()
```

```python
result = adfuller(sales_diff)

print("ADF Statistic:", result[0])
print("p-value:", result[1])
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

fig, ax = plt.subplots(1, 2, figsize=(15, 5))

plot_acf(sales_diff, lags=40, ax=ax[0])
plot_pacf(sales_diff, lags=40, ax=ax[1])

plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA
import warnings

warnings.filterwarnings("ignore")

best_aic = float("inf")
best_order = None

for p in range(4):
    for q in range(4):
        try:
            model = ARIMA(daily_sales, order=(p, 1, q))
            fitted_model = model.fit()

            if fitted_model.aic < best_aic:
                best_aic = fitted_model.aic
                best_order = (p, 1, q)

        except:
            continue

print("Best ARIMA Order:", best_order)
print("Best AIC:", best_aic)
```

```python
train_size = int(len(daily_sales) * 0.8)

train = daily_sales[:train_size]
test = daily_sales[train_size:]

print("Training data:", len(train))
print("Testing data:", len(test))
print("Training end date:", train.index[-1])
print("Testing start date:", test.index[0])
```
